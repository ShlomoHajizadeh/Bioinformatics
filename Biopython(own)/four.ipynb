```python
from Bio import SeqIO
import os

names = ("panTro6", "hg38", "rheMac10", "calJac4", "mm39", "rn7")
for i, name in enumerate(names):
    filename = f"/Users/shlomohajizadeh/Documents/GitHub/Bioinformatics/Biopython(own)/sequence.long.2bit"
    # Ensure the file exists at the specified path
    if not os.path.exists(name):
        print(
            f"Warning: File not found: {name}. Please ensure the file exists at the specified path or update the path to the correct location."
        )
        continue
    genome = SeqIO.to_dict(SeqIO.parse(filename, "twobit"))
    chromosome = genome_alignment.sequences[i].id
    assert len(genome_alignment.sequences[i]) == len(genome[chromosome])
    genome_alignment.sequences[i] = genome[chromosome]
    genome_alignment.sequences[i].id = f"{name}.{chromosome}"

print(genome_alignment)
```
